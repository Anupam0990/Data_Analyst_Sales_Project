Converting "Order_Date" to date format

```python
df["Order_Date"] = pd.to_datetime(df["Order_Date"])
```

```python
df.dtypes
```

Add a Column of Profit Margin

```python
df["Profit Margin"] = (df["Profit"]/df["Selling Price"])*100
```

```python
df.info()
```

Extracting Year and Month from Order_Date

```python
df["Year"] = df["Order_Date"].dt.year
df["Month"] = df["Order_Date"].dt.month
```

Calcualating Insights


```python
total_sales = df["Selling Price"].sum()
total_profit = df["Profit"].sum()
total_orders = df["Order_ID"].nunique()
total_customers = df["Customer_ID"].nunique()
total_quantity = df["Quantity"].sum()

average_order_value = total_sales / total_orders
profit_margin = (total_profit / total_sales) * 100

print("Total Sales:", total_sales)
print("Total Profit:", total_profit)
print("Total Orders:", total_orders)
print("Total Customers:", total_customers)
print("Total Quantity:", total_quantity)
print("Average Order Value:", average_order_value)
print("Profit Margin:", profit_margin)
```

```python
monthly_sales = (
    df.groupby(df["Order_Date"].dt.to_period("M")).agg(
          Sales=("Selling Price", "sum"),
          Profit=("Profit", "sum"),
          Orders=("Order_ID", "nunique")
      )
)
monthly_sales
```

Sales wise top 5 month

```python
monthly_sales.sort_values("Sales", ascending=False).head(5)
```

Sales wise bottom 5 month

```python
monthly_sales.sort_values("Sales").head(5)
```

Categorical Analysis

```python
category_analysis = (
    df.groupby("Category").agg(
          Sales=("Selling Price", "sum"),
          Profit=("Profit", "sum"),
          Orders=("Order_ID", "nunique"),
          Quantity=("Quantity", "sum")
      )
)

category_analysis["Profit_Margin"] = (
    category_analysis["Profit"] /
    category_analysis["Sales"] * 100
)
```


Sales wise top 5 category

```python
category_analysis.sort_values("Sales", ascending=False).head(5)
```

Sales Wise bottom 5 category

```python
category_analysis.sort_values("Sales").head(5)
```

ProfitMargin wise top 5 category

```python
category_analysis.sort_values("Profit_Margin", ascending=False).head(5)
```

ProfitMargin wise bottom 5 category

```python
category_analysis.sort_values("Profit_Margin").head(5)
```

Product Analysis

```python
product_analysis = (
    df.groupby("Product")
      .agg(
          Sales=("Selling Price", "sum"),
          Profit=("Profit", "sum"),
          Quantity=("Quantity", "sum"),
          Orders=("Order_ID", "nunique")
      )
)

product_analysis["Profit_Margin"] = (
    product_analysis["Profit"] /
    product_analysis["Sales"] * 100
)
```

Sales wise to 10 products

```python
product_analysis.sort_values("Sales", ascending=False).head(10)
```

Sales wise bottom 10 products

```python
product_analysis.sort_values("Sales").head(10)
```

```python
top_products = (
    product_analysis["Sales"].sort_values(ascending=False).head(10)   
)
```

Customer Analysis

```python
customer_analysis = (
    df.groupby("Customer_ID").agg(
          Orders=("Order_ID", "nunique"),
          Sales=("Selling Price", "sum"),
          Profit=("Profit", "sum"),
          Quantity=("Quantity", "sum")
      )
)
```

Sales wise top 10 customer ID

```python
customer_analysis.sort_values("Sales", ascending=False).head(10)
```

Sales wise bottom 10 customer ID

```python
customer_analysis.sort_values("Orders", ascending=False).head(10)
```

State Analysis

```python
state_analysis = (
    df.groupby("State")
      .agg(
          Sales=("Selling Price", "sum"),
          Profit=("Profit", "sum"),
          Orders=("Order_ID", "nunique"),
          Customers=("Customer_ID", "nunique")
      )
)

state_analysis["Profit_Margin"] = (
    state_analysis["Profit"] /
    state_analysis["Sales"] * 100
)
```

Sales wise top 10 states

```python
state_analysis.sort_values("Sales", ascending=False).head(10)
```

Sales wise bottom 10 states


```python
state_analysis.sort_values("Profit_Margin", ascending=False).head(10)
```

Return Analysis

```python
return_analysis = (
    df.groupby("Category")
      .agg(
          Orders=("Order_ID", "count"),
          Returns=("Returned", lambda x: (x == "Yes").sum())
      )
)

return_analysis["Return_Rate"] = (
    return_analysis["Returns"] /
    return_analysis["Orders"] * 100
)

return_analysis.sort_values(
    "Return_Rate",
    ascending=False
)
```

Generating Charts

```python
monthly_sales["Sales"].plot(figsize=(12,5), marker="o")

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig("monthly_sales.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
category_analysis["Sales"].sort_values().plot(
    kind="barh",
    figsize=(10,5)
)

plt.title("Sales by Category")
plt.xlabel("Sales")
plt.ylabel("Category")
plt.tight_layout()
plt.savefig("sales_by_category.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
state_analysis["Sales"].sort_values().plot(
    kind="barh",
    figsize=(10,7)
)

plt.title("Sales by State")
plt.xlabel("Sales")
plt.ylabel("State")
plt.tight_layout()
plt.savefig("Sales_By_State.png", dpi=300, bbox_inches="tight")
plt.show()
```
